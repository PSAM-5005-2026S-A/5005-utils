```python
import re
from os import listdir
from shutil import move
```

```python
DIR = "mixtape06"
fnames = sorted(f for f in listdir(DIR) if f.endswith(".mp3"))

for f in fnames:
  fi_path = f"{DIR}/{f}"
  fout = re.sub(r"[_]{2,}", "_", re.sub(r"\W", "_", f))
  fo_path = f"{DIR}/{fout}"

  move(fi_path, fo_path)
```
